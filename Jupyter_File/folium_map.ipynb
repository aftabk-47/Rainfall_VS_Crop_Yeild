```python
import json
import folium

# -------------------------------
# LOAD DATA
# -------------------------------
with open("state_prod_rain_fixed1.json", "r") as f:
    state_summary = json.load(f)

with open("india_states.geojson", "r", encoding="utf-8") as f:
    india_geo = json.load(f)

# -------------------------------
# DEBUG: CHECK MATCHING STATES
# -------------------------------
geo_states = [f["properties"]["NAME_1"].strip().lower() for f in india_geo["features"]]
json_states = list(state_summary.keys())

print("GeoJSON states sample:", geo_states[:5])
print("JSON states sample:", json_states[:5])

# -------------------------------
# NORMALIZE GEOJSON
# -------------------------------
for feature in india_geo["features"]:
    feature["id"] = feature["properties"]["NAME_1"].strip().lower()

# -------------------------------
# PREPARE DATA (SAFE)
# -------------------------------
choropleth_data = []

for state, stats in state_summary.items():
    state_id = state.strip().lower()

    if state_id in geo_states:   # ✅ ensures valid mapping
        choropleth_data.append([
            state_id,
            stats.get("Total Production", 0)
        ])

print("Mapped states:", len(choropleth_data))

# -------------------------------
# CREATE MAP
# -------------------------------
m = folium.Map(location=[23.5, 80.5], zoom_start=5)

# -------------------------------
# ADD CHOROPLETH
# -------------------------------
folium.Choropleth(
    geo_data=india_geo,
    data=choropleth_data,
    columns=[0, 1],
    key_on="feature.id",
    fill_color="YlGn",
    legend_name="Production"
).add_to(m)

# -------------------------------
# SAVE
# -------------------------------
m.save("map.html")

print("Map saved as map.html")
```
